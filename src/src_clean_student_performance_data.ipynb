```python
#Review unique values in categorical columns
print(df["gender"].unique())
print(df["race/ethnicity"].unique())
print(df["parental level of education"].unique())
print(df["lunch"].unique())
print(df["test preparation course"].unique())
```

```python
#Count the number of unique values in each test score column
print("Unique Math Scores:", df["math score"].nunique())
print("Unique Reading Scores:", df["reading score"].nunique())
print("Unique Writing Scores:", df["writing score"].nunique())
```

```python
#Check the minimum and maximum values for each test score
print("Math:", df["math score"].min(), "-", df["math score"].max())
print("Reading:", df["reading score"].min(), "-", df["reading score"].max())
print("Writing:", df["writing score"].min(), "-", df["writing score"].max())
```

```python
#Check for test scores outside the valid 0–100 range
df[
(df["math score"] < 0) | (df["math score"] > 100) |
(df["reading score"] < 0) | (df["reading score"] > 100) |
(df["writing score"] < 0) | (df["writing score"] > 100)
]
```

```python
#Check all text columns for leading or trailing spaces
df.select_dtypes(include="object").apply(
lambda column: column.astype(str).str.strip().ne(column.astype(str)).any()
)
```

```python
#Check for extra spaces in categorical columns
categorical_columns = df.select_dtypes(include="object").columns
for column in categorical_columns:
  has_extra_spaces = (
      df[column].dropna() != df[column].dropna().str.strip()
      ).any()
print(column, ":", has_extra_spaces)
```

```python
#Create a copy of the raw data for preprocessing
df_clean = df.copy()
```

```python
#Confirm the cleaned dataset
print("Raw dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)
print("\nMissing values:")
print(df_clean.isnull().sum())
print("\nDuplicate rows:", df_clean.duplicated().sum())
```

```python
#Save the cleaned student performance dataset
df_clean.to_csv("students_performance_clean.csv", index=False)
```
